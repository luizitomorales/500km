```python
# Pacote que fará o cálculo das distâncias em função das coordenadas

from geopy.distance import geodesic
```

```python
cidades
```

```python
#Função que dará essa distância em quilometros

def calc_dist(coord1, coord2):
    return geodesic(coord1, coord2).kilometers
    resultados= []
```

```python
# Iteração que relaciona cada uma das cidades aos aeroportos dos seus Estados e insere a coluna das distâncias

resultados= []
for idx_cidade, cidade in cidades.iterrows():
    uf_cidade = cidade['Sigla']
    for idx_aeroporto, aeroporto in aero[aero['UF'] == uf_cidade].iterrows():
        distancia = calc_dist((cidade['Latitude'], cidade['Longitude']), (aeroporto['Latitude'], aeroporto['Longitude']))
        resultados.append({
            'cidade': cidade['nome'],
            'UF da cidade': cidade['Sigla'],
            'Latitude da cidade':cidade['Latitude'],
            'Longitude da cidade': cidade['Longitude'],
            'Aeroporto': aeroporto['NOME'],
            'UF do Aeroporto': aeroporto['UF'],
            'Latitude do Aeroporto': aeroporto['Latitude'],
            'Longitude do Aeroporto': aeroporto['Longitude'],
            'SITUADO NA AMAZÔNIA LEGAL': aeroporto['SITUADO NA AMAZÔNIA LEGAL'],
            'Distância (em KM)': distancia
        })
```

```python
# Planilha resultante

resultado_final = pd.DataFrame(resultados)
```

```python
import mysql.connector
import csv

# Configurando a conexão ao banco de dados
config = {
    'user': 'root',
    'password': '',
    'host': 'localhost',
    'database': '500km'
}

try:
    conexao = mysql.connector.connect(**config)
    cursor = conexao.cursor()

    csv_caminho = 'dist_cidades_aero.csv'

    with open(csv_caminho, 'r', encoding='utf-8') as csvfile:
        reader = csv.reader(csvfile)
        next(reader)  # Pula a primeira linha (cabeçalho)

        for row in reader:
            cursor.execute('''
                INSERT INTO dist_cidades_aero (
                    cidade, `UF da cidade`, `Latitude da cidade`, `Longitude da cidade`,
                    Aeroporto, `UF do Aeroporto`, `Latitude do Aeroporto`,
                    `Longitude do Aeroporto`, `SITUADO NA AMAZÔNIA LEGAL`,
                    `Distância (em KM)`
                ) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
            ''', row)

    conexao.commit()

    cursor.close()
    conexao.close()

    print("Dados inseridos com sucesso!")

except mysql.connector.Error as err:
    print(f"Erro: {err}")
```
